## Тема: Проверка матрицы Грама

**Раздел книги:** Аналитическая геометрия.  
**Математическая тема:** проверка скалярного произведения через матрицу Грама; условия симметричности и положительной определённости матрицы $A$.

## Условие
Рассмотрим $\mathbb{R}^2$ с операцией $\langle \cdot, \cdot \rangle$, определённой как:
$$
\langle \mathbf{x}, \mathbf{y} \rangle := \mathbf{x}^\top A \mathbf{y}, \quad \text{где } A = \begin{bmatrix} 2 & 0 \\ 1 & 2 \end{bmatrix}.
$$
Является ли эта операция скалярным произведением?

## Математическая идея
Для того чтобы билинейная форма $\langle \mathbf{x}, \mathbf{y} \rangle = \mathbf{x}^\top A \mathbf{y}$ была скалярным произведением, матрица $A$ должна быть **симметричной** ($A = A^\top$) и **положительно определённой**.

Если $A$ не симметрична, то нарушается аксиома $\langle \mathbf{x}, \mathbf{y} \rangle = \langle \mathbf{y}, \mathbf{x} \rangle$.

## Решение

**Шаг 1. Проверка симметричности.**
Матрица $A = \begin{bmatrix} 2 & 0 \\ 1 & 2 \end{bmatrix}$.
Транспонированная матрица $A^\top = \begin{bmatrix} 2 & 1 \\ 0 & 2 \end{bmatrix}$.
Так как $A \neq A^\top$ (элементы $a_{12}=0$ и $a_{21}=1$ не совпадают), матрица **не является симметричной**.

**Шаг 2. Проверка аксиомы на векторах.**
Возьмём $\mathbf{x} = [1, 0]^\top$ и $\mathbf{y} = [0, 1]^\top$.
$$
\langle \mathbf{x}, \mathbf{y} \rangle = \begin{bmatrix} 1 & 0 \end{bmatrix} \begin{bmatrix} 2 & 0 \\ 1 & 2 \end{bmatrix} \begin{bmatrix} 0 \\ 1 \end{bmatrix} = \begin{bmatrix} 2 & 0 \end{bmatrix} \begin{bmatrix} 0 \\ 1 \end{bmatrix} = 0.
$$
$$
\langle \mathbf{y}, \mathbf{x} \rangle = \begin{bmatrix} 0 & 1 \end{bmatrix} \begin{bmatrix} 2 & 0 \\ 1 & 2 \end{bmatrix} \begin{bmatrix} 1 \\ 0 \end{bmatrix} = \begin{bmatrix} 1 & 2 \end{bmatrix} \begin{bmatrix} 1 \\ 0 \end{bmatrix} = 1.
$$
Поскольку $0 \neq 1$, свойство коммутативности нарушено.

**Итоговый ответ:** Данная операция **не является** скалярным произведением из-за несоблюдения свойства симметричности.

```python
# Матрица из условия
import numpy as np

A = np.array([[2, 0],
             [1, 2]])

x = np.array([1, 0])
y = np.array([0, 1])

# Вычисление "скалярного" произведения в обоих порядках
xy = x.T @ A @ y
yx = y.T @ A @ x

print(f"<x, y> = {xy}")
print(f"<e, x> = {yx}")
print(f"Симметрично? {np.isclose(xy, yx)}")
```

## Дополнительный пример: Симметризация формы

**Идея.** Любую билинейную форму можно разложить на симметричную и антисимметричную части. Скалярное произведение — это только симметричная часть. Если мы попробуем "починить" матрицу $A$, взяв от неё только симметричную часть $S = \frac{A + A^\top}{2}$, то получим уже корректное определение:
$$
S = \begin{bmatrix} 2 & 0.5 \\ 0.5 & 2 \end{bmatrix}
$$
Эта матрица $S$ симметрична и положительно определена, так что она действительно задаёт скалярное произведение. Это часто используется в ML при работе с косыми метриками или асимметричными графами связей.

```python
import numpy as np

# Симметричная часть матрицы A
S = (A + A.T) / 2
print("Симметризированная матрица S:\n", S)

# Теперь для <x, y>_S = x.T @ S @ y симметрия будет выполнена
xy_s = x.T @ S @ y
yx_s = y.T @ S @ x
print(f"Проверка симметри после исправления: {xy_s == yx_s}")
```

## Проверка результата

Мы доказали, что исходная матрица $A$ не является симметричной, что автоматически исключает возможность её использования в качестве тензора скалярного произведения. Численная проверка на конкретных единичных векторах подтвердила, что $\langle \mathbf{x}, \mathbf{y} \rangle \neq \langle \mathbf{y}, \mathbf{x} \rangle$. В дополнительном примере мы показали, как замена $A$ на её симметричную часть $S$ устраняет эту проблему.

## Вывод

Не всякая линейная комбинация координат двух векторов является скалярным произведением. Ключевое требование — симметрия матрицы весов $A$. В задачах оптимизации и машинного обучения использование несимметричных матриц вместо скалярного произведения может привести к тому, что понятие "длина вектора" или "угол между векторами" потеряет физический смысл.
