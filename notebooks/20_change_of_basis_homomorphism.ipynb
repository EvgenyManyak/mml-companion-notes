# Пример 20. Гомоморфизм со сменой базиса

## Тема

**Раздел книги:** Линейная алгебра.  
**Математическая тема:** матрица перехода между базисами $P=(\text{цель})^{-1}(\text{источник})$, матрица гомоморфизма в паре базисов и правило $A'=P_2\,A\,P_1$; различение эндоморфизма (подобие $P^{-1}AP$, ср. 19) и отображения между разными пространствами (эквивалентность $P_2 A P_1$); инвариантность ранга при умножении на обратимые матрицы.

## Условие
Даны векторы $\mathbb{R}^2$ в стандартном базисе:
$$
b_1=\begin{bmatrix}2\\1\end{bmatrix},\ b_2=\begin{bmatrix}-1\\-1\end{bmatrix},\ b_1'=\begin{bmatrix}2\\-2\end{bmatrix},\ b_2'=\begin{bmatrix}1\\1\end{bmatrix},
$$
и два упорядоченных базиса $B=(b_1,b_2)$, $B'=(b_1',b_2')$. 

**(a)** Докажите, что $B$ и $B'$ — базисы $\mathbb{R}^2$, нарисуйте их.

**(b)** Найдите матрицу $P_1$, меняющую базис с $B'$ на $B$.

**(c)** Для $c_1=\begin{bmatrix}1\\2\\-1\end{bmatrix}, c_2=\begin{bmatrix}0\\-1\\2\end{bmatrix}, c_3=\begin{bmatrix}1\\0\\-1\end{bmatrix}$ и $C=(c_1,c_2,c_3)$:

    (1) докажите базисность $C$ через детерминанты;
 
    (2) найдите матрицу $P_2$, меняющую базис с $C$ на стандартный $C'$.
 
**(d)** Гомоморфизм $\Phi:\mathbb{R}^2\to\mathbb{R}^3$ задан условиями $\Phi(b_1+b_2)=c_2+c_3$, $\Phi(b_1-b_2)=2c_1-c_2+3c_3$. Найдите матрицу $A_\Phi$ в парах базисов $B,C$.

**(e)** Найдите матрицу $A'$ в парах $B',C'$.

**(f)** Для $x$ с координатами $[2,3]^\top$ в $B'$:

    (1) координаты $x$ в $B$; 
 
    (2) координаты $\Phi(x)$ в $C$;
 
    (3) $\Phi(x)$ в стандартном базисе;
 
    (4) тот же результат через $[x]_{B'}$ и $A'$.

## Математическая идея
Если матрица $M_B$ составлена из столбцов базиса $B$ (в стандартных координатах), то координаты вектора связаны как $x_{std}=M_B[x]_B$. Отсюда **матрица перехода** «из $B'$ в $B$» (переводит $[x]_{B'}\mapsto[x]_B$):
$$
P_1=M_B^{-1}M_{B'}.
$$
Для отображения $\Phi$ с матрицей $A$ в паре $(B,C)$ и матрицей $A'$ в паре $(B',C')$ связывающее правило:
$$
A' = P_2\,A\,P_1,
$$
где $P_1$ меняет базис домена ($B'\!\to\!B$), а $P_2$ — кодомена ($C\!\to\!C'$). Это **эквивалентность** матриц (в отличие от подобия $P^{-1}AP$ для эндоморфизма из 19); ранг при этом инвариантен.

## Решение

**(a)** $\det M_B=\det\begin{bmatrix}2&-1\\1&-1\end{bmatrix}=-1\neq0$, $\det M_{B'}=\det\begin{bmatrix}2&1\\-2&1\end{bmatrix}=4\neq0$ ⇒ оба набора из 2 независимых векторов в $\mathbb{R}^2$ — базисы.

**(b)** $M_B^{-1}=\begin{bmatrix}1&-1\\1&-2\end{bmatrix}$, $P_1=M_B^{-1}M_{B'}=\begin{bmatrix}1&-1\\1&-2\end{bmatrix}\begin{bmatrix}2&1\\-2&1\end{bmatrix}=\begin{bmatrix}4&0\\6&-1\end{bmatrix}$.

**(c.1)** $\det M_C=\det\begin{bmatrix}1&0&1\\2&-1&0\\-1&2&-1\end{bmatrix}=1(1)+1(4-1)=4\neq0$ ⇒ $C$ — базис.

**(c.2)** $M_{C'}=I$, поэтому $P_2=M_{C'}^{-1}M_C=M_C=\begin{bmatrix}1&0&1\\2&-1&0\\-1&2&-1\end{bmatrix}$.

**(d)** Из двух условий, складывая/вычитая: $\Phi(b_1)=c_1+2c_3$, $\Phi(b_2)=-c_1+c_2-c_3$. Столбцы $A_\Phi$ — координаты в $C$:
$$
A_\Phi=\begin{bmatrix}1&-1\\0&1\\2&-1\end{bmatrix}.
$$

**(e)** $A'=P_2 A P_1=\begin{bmatrix}1&0&1\\2&-1&0\\-1&2&-1\end{bmatrix}\begin{bmatrix}1&-1\\0&1\\2&-1\end{bmatrix}\begin{bmatrix}4&0\\6&-1\end{bmatrix}=\begin{bmatrix}0&2\\-10&3\\12&-4\end{bmatrix}$.

**(f)**
1. $[x]_B=P_1[2,3]^\top=[8,9]^\top$.
2. $[\Phi(x)]_C=A_\Phi[8,9]^\top=[-1,9,7]^\top$.
3. $\Phi(x)=-c_1+9c_2+7c_3=[6,-11,12]^\top$.
4. $A'[2,3]^\top=[6,-11,12]^\top$ ✓ — совпало с п.3.

```python
import numpy as np

# --- данные ---
b1, b2 = np.array([2, 1.]), np.array([-1, -1.])
b1p, b2p = np.array([2, -2.]), np.array([1, 1.])
MB, MBp = np.column_stack([b1, b2]), np.column_stack([b1p, b2p])
c1, c2, c3 = np.array([1, 2, -1.]), np.array([0, -1, 2.]), np.array([1, 0, -1.])
MC = np.column_stack([c1, c2, c3])

# (a) базисность через детерминанты
print("(a) det MB =", round(np.linalg.det(MB), 3), "| det MB' =", round(np.linalg.det(MBp), 3),
      "| det MC =", round(np.linalg.det(MC), 3))

# (b) P1 = MB^{-1} MB'  (B' -> B)
P1 = np.linalg.inv(MB) @ MBp
print("(b) P1 =\n", np.round(P1, 6))

# (c) P2 = MC  (C -> стандартный)
P2 = MC.copy()

# (d) A в парах (B, C): координаты Phi(b1), Phi(b2) в C
A = np.array([[1,- 1],[0, 1],[2, -1]], dtype=float)

# (e) A' = P2 A P1
A_prime = P2 @ A @ P1
print("(e) A' =\n", np.round(A_prime, 6))

# (f) сквозная проверка
x_Bp = np.array([2., 3.])
x_B  = P1 @ x_Bp                 # 1
phi_C = A @ x_B                  # 2
phi_std = MC @ phi_C             # 3  (координаты в C -> стандартные)
phi_std_direct = A_prime @ x_Bp  # 4
print("(f) [x]_B =", x_B, "| [Phi(x)]_C =", phi_C)
print("    Phi(x) std (п.3) =", phi_std, "| п.4 =", phi_std_direct,
      "| совпали:", np.allclose(phi_std, phi_std_direct))
```

## Дополнительный пример: матрица гомоморфизма как «проектор эмбеддингов»

**Идея.** Линейный слой нейросети $y=Wx$ — это гомоморфизм между пространствами признаков. Если вход и выход записаны в разных базисах (например, до и после PCA-whitening), то «физическая» матрица слоя меняется по правилу $W'=P_2 W P_1$ — ровно как $A'$ в п. (e). Это объясняет, почему один и тот же слой выглядит по-разному в разных системах координат, но сохраняет ранг и действие на данные. В матричной факторизации рекомендательных систем переход между базисами пользователей/товаров — та же эквивалентность $P_2 A P_1$.

```python
import numpy as np
rng = np.random.default_rng(3)
W  = rng.normal(size=(3, 2))                  # слой в «сырых» базисах
P1 = np.linalg.inv(rng.normal(size=(2, 2)))   # новый базис домена
P2 = rng.normal(size=(3, 3))                  # новый базис кодомена
Wp = P2 @ W @ P1
print("rank(W) =", np.linalg.matrix_rank(W), "== rank(W') =", np.linalg.matrix_rank(Wp))
```

## Визуализация базисов $B$ и $B'$ (часть a)

Две панели: на каждой из стандартного начала координат отложены векторы базиса. Видно, что оба набора — две независимые стрелки, то есть разлагают $\mathbb{R}^2$; $B$ (синий) и $B'$ (оранжевый) ориентированы по-разному, что и отражает нетождественность $P_1$.

```python
import numpy as np, matplotlib.pyplot as plt
fig, ax = plt.subplots(1, 2, figsize=(9, 4.5))
for a, (vecs, col, name) in zip(ax, [([b1, b2], 'steelblue', 'Базис B'),
                                  ([b1p, b2p], 'darkorange', "Базис B'")]):
    for v in vecs:
        a.annotate('', xy=v, xytext=(0, 0), arrowprops=dict(arrowstyle='->', color=col, lw=2))
    a.set_xlim(-3, 3); a.set_ylim(-3, 3); a.axhline(0, c='k', lw=.5); a.axvline(0, c='k', lw=.5)
    a.grid(alpha=.3); a.set_aspect('equal'); a.set_title(name)
plt.tight_layout(); plt.show()
```

## Проверка результата

В части (b) матрица $P_1$ проверена на конкретном векторе: $[x]_B=P_1[x]_{B'}$ даёт $[8,9]^\top$, что независимо подтверждено решением системы $x=\alpha b_1+\beta b_2$ для $x=2b_1'+3b_2'=[7,-1]^\top$. В части (e) матрица $A'$ вычислена двумя путями — через формулу $A'=P_2AP_1$ и через прямое построение столбцов $\Phi(b_1'),\Phi(b_2')$ в стандартных координатах ($[0,-10,12]^\top$ и $[2,3,-4]^\top$), — и оба дают одинаковый результат. В части (f) все четыре способа вычисления $\Phi(x)$ сошлись к $[6,-11,12]^\top$; дополнительно инвариант ранга подтверждает корректность цепочки эквивалентных матриц ($\operatorname{rank}A_\Phi=\operatorname{rank}A'=2$).
